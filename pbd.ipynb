```python
folder = 'shoulder/train_txt_labels'
# 폴더 안에 txt 파일들을 읽고, 첫번째 글자가 0인 파일들을 list로 저장
import os
file_list = []
for file in os.listdir(folder):
    if file.endswith('.txt'):
        with open(os.path.join(folder, file), 'r') as f:
            first_line = f.readline().strip()
            if first_line.startswith('0'):
                file_list.append(file)
print(file_list)
```

```python
# file_list 안에 있는 파일들을 읽어서, row마다 첫번째 글자가 0인 경우, 개수 세기
# 그 개수가 2개인 파일 list로 저장
count_list = []
for file in file_list:
    count = 0
    with open(os.path.join(folder, file), 'r') as f:
        for line in f:
            if line.strip().startswith('0'):
                count += 1
    if count == 2:
        count_list.append(file)
print(count_list)
```
